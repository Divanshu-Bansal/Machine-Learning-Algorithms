```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=1000, n_features=2, n_classes=2, random_state=42, n_clusters_per_class=2, n_redundant=0)
```

```python
X
```

```python
y
```

```python
pd.DataFrame(X)
```

```python
sns.scatterplot(
    x=pd.DataFrame(X)[0],
    y=pd.DataFrame(X)[1],
    hue=y
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,random_state=42)
```

## SVC - Linear Kernel

```python
from sklearn.svm import SVC
svc = SVC(kernel='linear')
```

```python
svc.fit(X_train,y_train)
```

```python
y_pred = svc.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print("accuracy:",accuracy_score(y_test,y_pred))
print("Confusion Matrix:")
print(confusion_matrix(y_test,y_pred))
```

```python
print("Classification Report:")
print(classification_report(y_test,y_pred))
```

## SVC - RBF

```python
rbf = SVC(kernel='rbf')
rbf.fit(X_train,y_train)
```

```python
y_pred1 = rbf.predict(X_test)
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test,y_pred1))
print("")
print("Classification Report:")
print(classification_report(y_test,y_pred1))
```

## SVC - Polynomial Kernel

```python
polynomial = SVC(kernel='poly')
polynomial.fit(X_train,y_train)
```

```python
y_pred2 = polynomial.predict(X_test)
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test,y_pred2))
print("")
print("Classification Report:")
print(classification_report(y_test,y_pred2))
```

## Hyperparameter Tunining

```python
from sklearn.model_selection import GridSearchCV
```

```python
params = dict(
    C = [1.0,0.1,0.01,0.001,10.0,100,1000],
    kernel = ['linear', 'poly', 'rbf'],
    gamma = ['scale', 'auto']
)
```

```python
params
```

```python
grid = GridSearchCV(SVC(), param_grid=params, refit=True, cv = 5)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred3 = grid.predict(X_test)
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test,y_pred3))
print("")
print("Classification Report:")
print(classification_report(y_test,y_pred3))
```
